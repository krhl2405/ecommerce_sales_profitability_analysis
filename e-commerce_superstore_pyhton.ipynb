### 3.Data Cleaning and Date Preparation

```python
# Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'])
```

```python
# Convert Shipping Date to datetime
df['ship_date'] = pd.to_datetime(df['ship_date'])
```

```python
# Create month and year fields
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.month_name()
```

```python
# Check the cleaned dataset
print("Final Shape:", df.shape)
df.head()
```

```python
# snake casting of columns

df.columns= df.columns.str.lower()
df.columns= df.columns.str.replace(' ','_')
```

```python
### 5. Connecting python to SQL server 

import pandas as pd
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus

server = r"localhost\SQLExpress"
database = "ecommerce"

connection_string = quote_plus(
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server};"
    "DATABASE=master;"
    "Trusted_Connection=yes;"
)

master_engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={connection_string}"
)

with master_engine.connect() as connection:
    connection.execution_options(isolation_level="AUTOCOMMIT").execute(
        text("""
        IF DB_ID('ecommerce') IS NULL
            CREATE DATABASE ecommerce
        """)
    )

print("Database is ready!")

connection_string = quote_plus(
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
)

engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={connection_string}"
)

with engine.connect() as connection:
    print("Connected to ecommerce successfully!")
```

```python
df.to_sql(
    "ecommerce_sales",
    engine,
    if_exists="replace",
    index=False
)

print("E-commerce data uploaded successfully!")
```

```python
result = pd.read_sql(
    "SELECT TOP 5 * FROM ecommerce_sales",
    engine
)

result
```
